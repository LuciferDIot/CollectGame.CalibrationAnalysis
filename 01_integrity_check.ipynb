# STEP 1 - Dataset Integrity & Coverage Validation

## Objective
Verify that the calibration dataset preserves structural, temporal, and experimental integrity across all gameplay modes and participants.
**This step is mandatory before any numeric processing.**

## What we are checking
1.  **Completeness**: Every `userId` must appear in **all detected `modeId`s**.
2.  **Data Validity**: No missing values in critical columns (`deathOccurredInWindow`).
3.  **Plausibility**: No negative values in metric columns.


## Quick Start


```python
import pandas as pd
import numpy as np
import os

DATA_FILE = os.path.join('data', 'processed', 'calibration_dataset.csv')

# Load Data
if os.path.exists(DATA_FILE):
    df = pd.read_csv(DATA_FILE)
    print(f"Loaded Dataset: {len(df)} rows from {DATA_FILE}")
else:
    print(f"ERROR: File not found at {DATA_FILE}. Please run generate_calibration_dataset.ipynb first.")
```

## 1. Dynamic Discovery
We automatically detect which Modes and Users exist in the dataset. This ensures the code adapts to future changes (e.g., adding a 'Tutorial' mode).


```python
if 'modeId' in df.columns:
    modes = df['modeId'].unique()
    print(f"Detected Modes ({len(modes)}): {modes}")
else:
    print("CRITICAL ERROR: 'modeId' column missing!")

if 'userId' in df.columns:
    users = df['userId'].unique()
    print(f"Detected Users ({len(users)})")
else:
    print("CRITICAL ERROR: 'userId' column missing!")
```

## 2. Coverage Validation
**Requirement**: Every `userId` must have at least one record in all `modeId`s.
Missing a mode suggests data loss or a dropped connection during that session.


```python
print("--- Checking Mode Coverage ---")
missing_modes = []

for u in users:
    user_data = df[df['userId'] == u]
    user_modes = user_data['modeId'].unique()
    
    # Check set difference: Expected Modes - User's Modes
    diff = set(modes) - set(user_modes)
    if diff:
        missing_modes.append((u, diff))

if missing_modes:
    print(f"FAIL: {len(missing_modes)} users are missing modes!")
    for u, m in missing_modes:
        print(f"  User {u} missing: {m}")
else:
    print(f"PASS: All {len(users)} users have data for all {len(modes)} detected modes.")
```

## 3. Data Completeness & Validity
**Requirement**:
*   `deathOccurredInWindow` must never be NaN.
*   `deathCountInWindow` must be non-negative.
*   Metrics should generally be non-negative.


```python
print("\n--- Checking Data Validity ---")

# 1. Critical Columns Check
critical_cols = ['deathOccurredInWindow', 'deathCountInWindow']
if df[critical_cols].isnull().any().any():
     print("FAIL: Found NaNs in critical columns!")
     print(df[critical_cols].isnull().sum())
else:
     print("PASS: No missing values in critical derived columns.")

# 2. Consistency Check
# deathOccurred == 1 IFF deathCount >= 1
inconsistent = df [
    ((df['deathOccurredInWindow'] == 1) & (df['deathCountInWindow'] == 0)) |
    ((df['deathOccurredInWindow'] == 0) & (df['deathCountInWindow'] > 0))
]
if len(inconsistent) > 0:
    print(f"FAIL: Found {len(inconsistent)} inconsistent death flags!")
else:
    print("PASS: Death indicators are consistent.")

# 3. Negative Value Check (Heuristic)
# Exclude ID and Timestamp columns from this check
non_metric_cols = ['userId', 'modeId', 'timestamp', 'time', '_id', 'time_dt', 'window_time_key']
numeric_cols = df.select_dtypes(include=[np.number]).columns
metric_cols = [c for c in numeric_cols if c not in non_metric_cols]

neg_counts = (df[metric_cols] < 0).sum()
if neg_counts.sum() > 0:
    print("WARNING: Found negative values in possible metrics:")
    print(neg_counts[neg_counts > 0])
else:
    print(f"PASS: No negative values found in {len(metric_cols)} metric columns.")
```
